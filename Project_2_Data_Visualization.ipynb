```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python

# Load the Tips dataset

tips = sns.load_dataset("tips")

# Display first five records
print("FIRST FIVE RECORDS")
print(tips.head())
```

```python

# Check dataset shape
print("DATASET SHAPE:")
print(tips.shape)

# Check column names
print("\nCOLUMN NAMES:")
print(tips.columns)

# Check missing values
print("\nMISSING VALUES:")
print(tips.isnull().sum())

# Statistical summary
print("\nSTATISTICAL SUMMARY:")
print(tips.describe())
```

```python

# HISTOGRAM

plt.figure(figsize=(8, 5))

sns.histplot(tips["total_bill"], bins=20)

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill (USD)")
plt.ylabel("Number of Meals")

plt.show()
```

```python

# BOX PLOT

plt.figure(figsize=(8, 5))

sns.boxplot(data=tips, x="day", y="tip")

plt.title("Tips Distribution by Day")
plt.xlabel("Day")
plt.ylabel("Tip Amount (USD)")

plt.show()
```

```python

# SCATTER PLOT

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)

plt.title("Relationship Between Total Bill and Tip")
plt.xlabel("Total Bill (USD)")
plt.ylabel("Tip Amount (USD)")

plt.show()
```


TAKEAWAY:
There is a positive relationship between total bill and tip amount. As the total bill increases, the tip generally increases.


```python

# CORRELATION HEATMAP

plt.figure(figsize=(8, 5))

correlation = tips.select_dtypes("number").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues",
    linewidths=0.5
)

plt.title("Correlation Between Numerical Variables")

plt.show()
```


TAKEAWAY:
Total Bill and Tip have the strongest positive correlation of 0.68. This indicates that higher restaurant bills generally tend to be associated with higher tips.


```python

# FINAL DATA VISUALIZATION DASHBOARD

sns.set_theme(style="whitegrid", palette="deep")

# Create 4 chart spaces
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# 1. Histogram
sns.histplot(tips["total_bill"], bins=20, ax=axes[0, 0])
axes[0, 0].set(
    title="Distribution of Total Bills",
    xlabel="Total Bill (USD)",
    ylabel="Number of Meals"
)

# 2. Box Plot
sns.boxplot(data=tips, x="day", y="tip", ax=axes[0, 1])
axes[0, 1].set(
    title="Tips by Day",
    xlabel="Day",
    ylabel="Tip (USD)"
)

# 3. Scatter Plot
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time",
    ax=axes[1, 0]
)
axes[1, 0].set(
    title="Bigger Bills Bring Bigger Tips",
    xlabel="Total Bill (USD)",
    ylabel="Tip (USD)"
)

# 4. Correlation Heatmap
sns.heatmap(
    tips.select_dtypes("number").corr(),
    annot=True,
    cmap="Blues",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Correlation Between Numbers")

# Adjust chart spacing
plt.tight_layout()

# Create outputs folder if needed
import os
os.makedirs("outputs", exist_ok=True)

# Save dashboard
plt.savefig("outputs/tips_overview.png", dpi=150)

# Display dashboard
plt.show()

print("Dashboard saved successfully!")
```
